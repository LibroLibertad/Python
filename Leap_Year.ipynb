A year with 366 days is called a leap year. A year is a leap year if it is divisible by 4. However, since the introduction of the Gregorian calendar in 1582, a year is not a leap year if it is divisible by 100; however, it is a leap year if it is divisible by 400. Write a program that asks the user for a year and computes whether that year is a leap year. 

Use the following code to get the year from the user:

year = int(input("Enter a year: "))

```python
def isLeapYear():
    year = int(input("Enter a year: "))
    # Your code here
    if(year % 400 == 0):
        return True; # OR return "This is a leap year.";
    elif(year % 100 == 0):
        return False; # OR return "This is not a leap year.";
    elif(year % 4 == 0):
        return True; # OR return "This is a leap year.";
    else:
        return False; # OR return "This is not a leap year.";
```

```python
isLeapYear()
```
